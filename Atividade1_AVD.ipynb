### Montar um script para fazer a extração para as séries 

```python
import requests

# URL da API do IBGE
url = (
    "https://servicodados.ibge.gov.br/api/v3/agregados/4093/"
    "periodos/all/variaveis/4099|4096|12466"
    "?localidades=N3[26]&classificacao=2[6794,4,5]"
)

# requisição para a API
response = requests.get(url)

# Transformando a resposta em JSON
data = response.json()

# Percorrendovariáveis retornadas pela API
for variavel in data:

    print("------------------------------------------------------------------------------------------------")
    print("Variável:", variavel["variavel"])
    print("------------------------------------------------------------------------------------------------")

    # Percorrendo os resultados de cada variável
    for resultado in variavel["resultados"]:

        # Identificando o sexo
        sexo = list(resultado["classificacoes"][0]["categoria"].values())[0]

        # Pegando a série histórica
        serie = resultado["series"][0]["serie"]

        print("Sexo:", sexo)

        # Mostrando período e valor
        for periodo, valor in serie.items():
            print(periodo, "-", valor)

        print()
```
