```python
import requests

# API da base de atendimentos do SAMU
url = (
    "https://dados.recife.pe.gov.br/api/action/datastore_search"
    "?resource_id=d6f586c3-34c5-4414-9804-6fdfbd75c7db"
    "&limit=200000"
)

# Requisição para a API
response = requests.get(url)

# Convertendo a resposta para JSON
data = response.json()

# Extraindo os registros
registros = data["result"]["records"]

print("Total de registros extraídos:", len(registros))
print()

# Exibindo os primeiros registros
for registro in registros[:10]:
    print(registro)
    print()
```
